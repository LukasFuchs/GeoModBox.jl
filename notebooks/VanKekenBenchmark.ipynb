# Van Keken Benchmark

Calculate the density difference between the upper and lower layer

$\begin{equation}
Rb = \frac{\Delta \rho g h^3}{\kappa \eta_r}
\end{equation}$

$\begin{equation}
\begin{gather*}
& g = 10.0 \textrm{ [m/s²] } \\
& h = 4000.0 \textrm{ [m] } \\
& \kappa = 10^{-6} \textrm{ [m²/s] } \\
& \eta_r = 10^{20} \textrm{ [Pa} \cdot \textrm{s] }
\end{gather*}
\end{equation}$

```julia
g = 10; h = 4000; κ = 1e-6; ηᵣ = 1e20; Rb = 1

Δρ = Rb * κ * ηᵣ / (g * h^3)

println("Δρ = ",Δρ ," kg/m³")
```
